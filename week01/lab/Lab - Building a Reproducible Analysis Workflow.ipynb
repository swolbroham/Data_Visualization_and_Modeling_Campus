## Step 3: Operation

Create one Boolean condition that identifies Canaveral National Seashore and another that identifies July.


```python
# Write your code here.
parkCanaveral = park_visits["park"] == "Canaveral National Seashore"
monthJuly = park_visits["month_number"] == 7
```

Combine the two conditions with `&`. Use `.loc` to select the `year` and `recreation_visits` features, make a copy named `canaveral_july`, and display it.


```python
# Write your code here.
canaveral_july = park_visits.loc[parkCanaveral & monthJuly][["year", "recreation_visits"]]

canaveral_july
```

Create `pre_pandemic_condition`, which is `True` for years from 2015 through 2019. Use it with `.loc` to calculate the mean of `recreation_visits`, and store the result as `pre_pandemic_july_mean`.

Display `pre_pandemic_july_mean`.


```python
# Write your code here.

pre_pandemic_condition = (park_visits["year"] <= 2019)&(park_visits["year"]>=2015) & monthJuly
pre_pandemic_july_mean = park_visits.loc[pre_pandemic_condition]["recreation_visits"].mean()

pre_pandemic_july_mean

# park_visits["pre_pandemic_condition"] = park_visits["year"].apply(lambda x: 2015 <= x <= 2019)
```

Create a condition that identifies rows from 2020 onward whose `recreation_visits` are greater than or equal to `pre_pandemic_july_mean`. Use `.loc` to create `recovery_years`, then display it in ascending year order.


```python
# Write your code here.
after_pandemic = park_visits["year"] > 2019
greaterThanprePandemic = park_visits["recreation_visits"] >= pre_pandemic_july_mean
recovery_years = park_visits.loc[monthJuly &greaterThanprePandemic & after_pandemic & parkCanaveral].sort_values("year",ascending = True)
recovery_years
```
