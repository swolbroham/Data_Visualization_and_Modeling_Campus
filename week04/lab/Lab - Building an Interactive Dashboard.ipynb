## 16. Create reusable figure code

Write a function that accepts the selected control value, filters or summarizes `analysis`, creates both Plotly figures, and returns the figures in a consistent order. Fix axis ranges or category orders when they need to remain stable across selections.



```python
def create_figures(selected_neighbourhood):

    # Keep listings in the selected neighborhood
    filtered = analysis[
        analysis["neighbourhood"] == selected_neighbourhood
    ]

    # Keep room types in the same order across selections
    room_order = [
        "Entire home/apt",
        "Private room",
        "Shared room",
        "Hotel room"
    ]

    # First chart: nightly prices
    fig_price = px.box(
        filtered,
        x="room_type",
        y="price",
        color="room_type",
        category_orders={"room_type": room_order},
        title=f"Airbnb Prices in {selected_neighbourhood}",
        template="plotly_white",
        height=550,
        labels={
            "room_type": "Room type",
            "price": "Nightly price (USD)"
        }
    )

    fig_price.update_layout(showlegend=False)
    fig_price.update_yaxes(range=[0, analysis["price"].max() * 1.05])

    # Second chart: availability
    fig_availability = px.box(
        filtered,
        x="room_type",
        y="availability_365",
        color="room_type",
        category_orders={"room_type": room_order},
        title=f"Airbnb Availability in {selected_neighbourhood}",
        template="plotly_white",
        height=550,
        labels={
            "room_type": "Room type",
            "availability_365": "Days available in the next 365 days"
        }
    )

    fig_availability.update_layout(showlegend=False)
    fig_availability.update_yaxes(range=[0, 365])

    return fig_price, fig_availability

```

```python
# previewing function
fig_price, fig_availability = create_figures(
    neighbourhood_options[0]["value"]
)

fig_price.show()
fig_availability.show()
```

## 17. Build the layout

Create a Dash app and its layout. Include the title, purpose statement, interactive control, two empty `dcc.Graph` components with unique IDs, and short conclusion and limitation statements.



```python
app = Dash(__name__)

app.layout = html.Div(
    [
        html.H1("Airbnb Prices and Availability in Los Angeles"),

        html.P(
            "Compare nightly prices and availability by room type "
            "to help plan your stay in Los Angeles."
        ),

        html.Label("Choose a neighborhood:", htmlFor="neighbourhood-dropdown"),

        dcc.Dropdown(
            id="neighbourhood-dropdown",
            options=neighbourhood_options,
            value=neighbourhood_options[0]["value"],
            clearable=False
        ),

        dcc.Graph(id="price-chart"),

        dcc.Graph(id="availability-chart"),

        html.H3("Conclusion"),
        html.P(
            "In Alhambra, entire homes and apartments generally have "
            "higher nightly prices than private or shared rooms. "
            "Private and shared rooms may offer more affordable options.",
            id="conclusion-text"
        ),

        html.H3("Limitation"),
        html.P(
            "The number of available days does not show whether a listing "
            "is available for your specific travel dates. Although each "
            "neighborhood has at least 300 listings, some room types may "
            "have few listings, making those comparisons less reliable."
        )
    ],
    style={
        "maxWidth": "1000px",
        "margin": "auto",
        "padding": "20px",
        "fontFamily": "Arial"
    }
)
```

## 18. Connect the callback

Connect the control's `value` to the `figure` property of both graphs. The callback function should call the reusable figure function and return its two figures in the same order as the two outputs.



```python

@app.callback(
    Output("price-chart", "figure"),
    Output("availability-chart", "figure"),
    Input("neighbourhood-dropdown", "value")
)
def update_charts(selected_neighbourhood):
    return create_figures(selected_neighbourhood)
```

## 19. Check the callback as a Python function

Call the callback or reusable figure function with at least two valid control values. Confirm that each call returns two Plotly figures without an error.



```python
for option in neighbourhood_options[:2]:
    neighbourhood = option["value"]

    fig_price, fig_availability = update_charts(neighbourhood)

    print(f"Neighborhood: {neighbourhood}")
    print(type(fig_price))
    print(type(fig_availability))

    fig_price.show()
    fig_availability.show()
```
